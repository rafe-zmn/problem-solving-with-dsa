
Problem: Contains Duplicate
Given an integer array nums, return true if any value appears at least twice in the array, and return false if every element is distinct.

Difficulty: Easy
###### Platform: LeetCode 217 (https://leetcode.com/problems/contains-duplicate/description/)

Solved the problem in three approch

*   Brute force → O(n²)
*   Sorting → O(n log n)
*   Set/hash → O(n)








```python
#Brute Force
#O(n^2)

def arr_duplicate(arr):
  n = len(arr)

  for i in range(n-1):
    for j in range(i+1, n):
      if arr[i] == arr[j]:
        return True
  return False

arr = [1,2,3,4,2]
arr_duplicate(arr)
```

```python
#sorting approch
#O(n log(n))

def dupFind(a):
  a.sort()

  for i in range(1, len(a)):
    if a[i] == a[i-1]:
      return True
  return False

a = [1,2,3,4,7,5]
print(dupFind(a))

```

```python
#hash approch
#O(n)

def dupFind2(num):
  seen = set()

  for i in range(len(num)):
    if num[i] in seen:
      return True
    seen.add(i)

  return False

n = [1,2,3,5,6,1]

print(dupFind2(n))

```
