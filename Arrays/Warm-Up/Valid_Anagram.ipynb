Problem: Valid Anagram

***Given two strings s and t, return true if t is an anagram of s, and false otherwise.***

Difficulty: Easy
###### Platform: LeetCode 242 (https://leetcode.com/problems/valid-anagram/description/)

Solved the problem in three approch

*   Sorting → O(n log n)
*   Frequency map → O(n)

```python
#Sorting
def is_anagram(a, b):
  return sorted(a) == sorted(b)

s = "anagram"
t = "nagaram"
print(is_anagram(s,t))
```

```python
#Frequency
def anagram(s,t):
  s, t = s.lower(), t.lower()

  if len(s) != len(t):
    return False

  s_count = {}
  t_count = {}

  for char in s:
    s_count[char] = s_count.get(char, 0) + 1

  for i in t:
    t_count[i] = t_count.get(i, 0) + 1

  return s_count == t_count


s = "aab"
t = "baA"
print(anagram(s,t))
```
